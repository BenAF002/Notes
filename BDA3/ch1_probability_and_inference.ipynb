# 1.1 The Three Steps of Bayesian Data Analysis

1. Setting up a **Full Probability Model**
    -  This is a joint distribution over all observable and unobservable quantities
2. Conditioning on Observed Data
    - Calculating the ***Posterior Distribution***
    - I.e. modeling the distribution of unobserved quantities given the observed data
3. Evaluating model fitness

# 1.2 General Notations

***Exchangeability***\
Random variabels are *exchangeable* if their joint probability density $p(y_1, ..., y_n)$ is *invariant* to permutations of their indexes. 

***Hierarchical Modeling***\
Used when information is available on "several different levels of observational units"

# 1.3 Bayesian Inference

#### Bayes' Rule

By the **Chain Rule of Probability** we may write the joint probability mass or density of $\theta$ and $y$ as: $$p(\theta, y) = p(y|\theta)p(\theta)$$
This is the product of the ***Sampling Distribution*** (or "Likelilhood") and the ***Prior***.

Using conditional probability, Baye's Rule is given as:
$$p(\theta| y) = \frac{p(y|\theta)p(\theta)}{p(y)} = \frac{p(y|\theta)p(\theta)}{\sum_\theta p(y|\theta)p(\theta)}$$

Omitting the denominator (which is often unknown) yields the proportional ***Unormalized Posterior Density***:
$$p(\theta | y) \propto p(\theta)p(y|\theta)$$

**The primary task** of any Bayesian approach is to model $p(\theta, y)$ and appropriately summarize $p(\theta | y)$

#### Prediction

Assuming a continuous $\theta$, the distribution of $y$ is:
$$p(y) = \int p(y, \theta)d\theta = \int p(y|\theta)p(\theta)d\theta$$
This is the marginal distribution of $y$ and is called the ***Prior Predictive Distribution***

After observing $y$, we can predict an unknown observable $\tilde y$ from the same process:
$$
\begin{align*}
& p(\tilde y | y) = \int p(\tilde y, \theta | y)d\theta \\
& \ \ \ \ \ \ \ \ \ \ \ = \int p(\tilde y | \theta , y) p(\theta | y) d\theta \\ 
& \ \ \ \ \ \ \ \ \ \ \ = \int p(\tilde y| \theta ) p(\theta | y)d\theta
\end{align*}
$$

This is called the ***Posterior Predictive Distribution*** and follows seemlessly from the chain rule.\
**NOTE**: The third line drops the conditional $y$. This is follows from a **crucial assumption** that $\tilde y$ and $y$ are **independent conditionally on $\theta$** 

#### Likelihood and Odds Ratios

$$ \frac{p(\theta_1|y)}{p(\theta_2|y)} = \frac{p(\theta_1)p(y|\theta_1)}{p(\theta_2)p(y|\theta_2)} $$
This handy expression is the ratio of posteriors for different $\theta$ values.\
**NOTE:** The ***Posterior Odds Ratio*** is equal to the prior odds multiplied by the ***Likelihood Ratio***
