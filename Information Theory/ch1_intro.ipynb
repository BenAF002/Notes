"Information theory answers two fundamental questions:"
- "What is the ultimate data compression?" (**Entropy:** $H$)
- "What is the ultimate transmission rate of communication" (**Channel Capacity:** $C$)

**Entropy** of a random variable $X$ with probability mass function (PMF) $p(x)$ is:
$$H(X) = -\sum_x p(x)\log_2 p(x)$$

Log base 2 is used by default. When log base 2 is used, the entropy is measured in ***Bits*** (guessing this will be explained later).

We can think of entropy as a measure of the *average uncertainty* in a random variable - or, as the *average number of bits* needed to describe the random variable. 

**But what is a "Bit"**?\
A bit is a boolean value ("yes/no", "heads/tails", $0/1$)\
We can think of the number of bits needed to represent an outcome or state as the number of yes/no questions needed to identify the state or as the length of a binary code representing the state.

It turns out that the entropy (when using log base 2) of a random variable is the average number of bits needed to describe the random variable - i.e. the average number of yes/no questions needed to identify which of the possibe states the random variable is in.\
For instance, an RV $X\sim U(1,32)$ can be in one of $32$ different states, each with equal probability. The entropy of $X$ is:
$$H(X) = -\sum_{i=1}^{32}\frac{1}{32}\log \frac{1}{32} = \log 32 = 5$$
This means that we can represent any of the $32$ possible states of the random variable with $5$ bits of information on average.\
$2^5 = 32$, so we may assign each of the $32$ outcomes to a 5-bit binary string. This means that 5-bits is both the *necessary* and *sufficient* number of bits to represent all states of the random variable.

Not all random variables have outcomes that are all representable by their entropy. Entropy is only guaranteed to be the *average* number of bits needed to represent a random variable's outcomes. When the entropy is *not* a power of 2, and when the distribution is *not uniform*, then some states may require fewer bits, and others may require more. The entropy is just the average.

*Another Conception*:\
A bit is a halving of a probability space. An observation with *one bit* of information, segments the probability space of an observable in half. Two bits, in quarters, three bits in eighths, and so on. \
The *probability* of a state is: $$\bigg(\frac{1}{2}\bigg)^I = p(x)$$
Where $I$ is the number of bits of $p(x)$.

The number of bits associated with an observation is **the number of times that you need to halve the possibility space** in order to **identify that observation**. ***BAAAMM***

**Conditional Entropy**\
Conditional entropy $H(X|Y)$ is the entropy of a random variable given knowledge of another random variable.\
The reduction in uncertainty due to another random variable is called the ***Mutual Information***, denoted:
$$I(X;Y) \coloneqq H(X) - H(X|Y) = \sum_{x,y} p(x,y) \log \frac{p(x,y)}{p(x)p(y)}$$
The mutual information between $X$ and $Y$ is **symmetric** in $X$ and $Y$, and is equal to $0$ if and only if $X$ and $Y$ are independent.

**Communication Channel**\
A communication channel is any system in which the output depnds "probabilistically" on its input.\
It can be characterized by a ***Probability Transition Matrix*** $p(y|x)$ that determines the conditional distribution of the output $Y$ over the input $X$.

The ***Channel Capacity*** is denoted $C$ and is given by: $$C = \max_{p(x)} I(X;Y)$$
In other words, the channel capacity $C$ is the maximum information gain over all possible input distributions.

**Relative Entropy**\
Relative entropy is the more generalized dual to mutual information. It is the "distance" between two PMFs:
$$D(p\|q) = \sum_x p(x)\log\frac{p(x)}{q(x)}$$
Note that $D(p\|q)$ is always non-negative and is $0$ only when $p(x) = q(x)$

We can use relative entropy to define a *geometry* for probability distributions.
