**Big Idea**:
$$P(x|\theta) \mapsto P(\theta|x)$$
Classic framework estimates likelihood of the observed data given model parameters (on the left).\
Bayesian framework estimates the likelihood of the parameters given the observed data (on the right).

**Bayes Theorem**: $$P(\theta |x) = \frac{P(x|\theta)P(\theta)}{P(x)} \equiv \text{Posterior} = \frac{\text{Likelihood} \times \text{Prior}}{\text{Sampling Distribution}}$$

**Pros**:
- Balances data $x$ with prior beliefs $p(\theta)$
- Gives distributions with real probabilities for $\theta$

**Cons**:
- Can be highly sensitive to prior and to subjective choices of priors
- Can get very computationally complex when the posterior is **not analytically tractable** (which is common)

## MLE OLS with Bayesian Priors

The MLE formulation of OLS, assuming homogeneous Gaussian errors is:
$$p(y|x, \theta) = \mathcal{N}(\theta x, \sigma^2)$$
With log-likelihood:
$$
\begin{align*}
& l(\theta) = \log\prod p(y_i|x_i, \theta) \\
& \ \ \ \ \ \ = \sum \log p(y_i|x_i, \theta) \\
& \ \ \ \ \ \ = n\log\bigg(\frac{1}{\sqrt{2\pi \sigma^2}}\bigg) -\frac{1}{2\sigma^2} \sum (y_i - \theta x_i)^2 \\ \ \\
& \ \ \implies l(\theta) \propto \|Y - \Theta X\|^2
\end{align*}
$$

## Maximum a Posteriori (MAP) Estimation

Start with: $$p(\theta | x, y) \propto p(y|x, \theta)p(\theta)$$
Goal is to estimate the log-posterior: $$\log \big( p(\theta | x, y) \big) = \log\big(p(y|x,\theta)\big) + \log\big(p(\theta)\big)$$
This decomposition is: $$\log\text{Posterior} = \log\text{MLE} + \log\text{Prior (regularizer)}$$

Suppose the prior is Gaussian: $$p(\theta) = \mathcal{N} (0, \sigma^2) = e^{-\theta^2/2\sigma^2}$$
Then:
$$\log\big(p(\theta|x,y)\big) = \|Y - \Theta X\|^2 + \|\Theta\|^2 \frac{1}{2\sigma^2} = \|Y - \Theta X\|^2 + \lambda \|\Theta\|^2 $$

This is *literally* just RIDGE regression

If the prior has a Laplace distribution then we get LASSO regression.
