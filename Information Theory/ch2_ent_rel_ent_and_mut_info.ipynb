# 2.1 Entropy

The ***alphabet*** of a randome variable $X$ is denoted $\mathcal X$. It is the set of all possible states of $X$

**Definition (Entropy)**:\
The entropy $H(X)$ of a discrete random variable $X$ is:
$$H(X) = -\sum_{x\in\mathcal X} p(x) \log_2 p(x)$$

The *information* of an observation $x$, measured in *bits* is:
$$I = -\log_2 p(x)$$
Thus, the entropy of a random variable is the *expected value* of its information over all its possible states (i.e. over $\mathcal X$).

# 2.2 Joint Entropy and Conditional Entropy

**Definition (Join Entropy)**:\
The joint entropy $H(X,Y)$ of a pair of discrete random variables is:
$$
\begin{align*}
& H(X,Y) = -\sum_{x\in\mathcal X}\sum_{y\in\mathcal Y} p(x,y) \log p(x,y) \\
& \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ = - E\log p(X, Y)
\end{align*}
$$

**Definition (Conditional Entropy)**:\
The conditional entropy $H(Y|X)$ for a pair of discrete random variables is:
$$
\begin{align*}
& H(Y|X) = \sum_{x\in\mathcal X} p(x) H(Y|X=x) \\
& \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ = - \sum_{x\in\mathcal X}\sum_{y\in\mathcal Y} p(x,y) \log p(y|x) \\
& \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ = -E\log p(Y|X)
\end{align*}
$$

**Theorem 2.2.1 (Chain Rule)**:
$$ H(X, Y) = H(X) + H(Y|X)$$
$$ \ \ \ \ \ \ \ \ E\log p(X,Y) = E\log p(X) + E\log p(Y|X)$$

**Chain Rule Corrolary**:
$$H(X, Y|Z) = H(X|Z) + H(Y|X, Z)$$

Both the chain rule of entropy and its corrolary are readily provable using the linearity of expectations and the chain rule of probability: $$p(A, B) = p(A|B)p(B)$$

# 2.3 Relative Entropy and Mutual Information

**Definition (Relative Entropy)**:\
Relative Entropy, aka. ***Kullback-Leibler Distance***, is the expected value of the log likelihood ratio between two distribution:
$$
\begin{align*}
& D(p\|a) = \sum_{x\in \mathcal X} p(x) \log \frac{p(x)}{q(x)} \\
& \ \ \ \ \ \ \ \ \ \ \ \ \ = E_p \log\frac{p(X)}{q(X)}
\end{align*}
$$
A way to interpret the relative entropy is as a measure of the inefficiency in modeling a distribution with $q$ when the *true* distribution is $p$.\
If we new the true distribution, then we could descibe $X$ with a code of $H(p)$ bits on average.\
However, if we only know $q$, then we need $H(p) + D(p\|q)$ bits on average.

**NOTE:**
- Relative entropy is always *non-negative*
- Relative entropy is $0$ if and only if $p = q$
- Relative entropy **need not be symmetric**

**Definition (Mutual Information)**:\
The mutual information $I(X;Y)$ between two random variables is the relative entropy between their joint distribution and the product of their marginal distributions:
$$
\begin{align*}
& I(X;Y) = \sum_{x\in\mathcal X} \sum_{y\in \mathcal Y} p(x,y) \log\frac{p(x,y)}{p(x)p(y)} \\
& \ \ \ \ \ \ \ \ \ \ \ \ \ \ = D(p(x,y) \| p(x) p(y))
\end{align*}
$$
