## Functionals

A function $y(x)$ maps an input value $x$ to an output value $y$: $$y(x): x \mapsto y$$

A ***Functional*** $F[y]$ maps a *function* $y(x)$ to an output value $F$:
$$F[y]: y(x) \mapsto F$$
For example, entropy $H[p(x)]$ is a functional. It maps a function $p(x)$ (a probability density function) to a scalar value: the entropy of $x$ under that density function.

## Calculus of Variations

How to find stationary points (minima/maxima) of functionals?\
Of all possible functions $y(x)$, which function maximizes or minimizes the funcitonal $F[y]$?

Consider how $F[y]$ changes in response to small changes $\epsilon \eta(x)$ to the function $y(x)$:
$$
F\big[y(x) + \epsilon \eta(x) \big] = F[y(x)] + \epsilon \int \frac{\delta F}{\delta y(x)} \eta(x) dx + \mathcal{O}(\epsilon^2)
$$

Where $\mathcal{O}(\epsilon^2)$ is all higher order terms in the Taylor expansion than the first. They're denoted this way because they become asymptotically small as $\epsilon \rightarrow 0$. The partial derivative of the functional w.r.t. $y$ is integrated over because $y(x)$ is infinitely dimensioned. In order to identify $y(x)$ from this space of all possible functions, we would need to know its realization at all values of $x$. So, infinitely many. Contrast this with a function from a well-defined finite-dimensional function space like the space of all polynomials of order $n$. Any such function is uniquely defined by its $n+1$ coefficients $(a_0, ..., a_n)$. Likewise, knowing $n+1$ realizations of the function, corresponding to $(x_0, ..., x_n)$ points in input space, would uniquely define the function. Then we'd sum over the first-order partial derivatives to get their combined contribution. For function spaces of infinite dimension, we need infinitely many input points to uniquely define the function. Thus, we integrate over first-order partial derivatives, since there are infinitely many contributions.

The functional is *stationary* iff:
$$\int \frac{\delta F}{\delta y(x)}\eta(x) dx = 0, \ \forall \ \eta(x)$$
